```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    'one': [12, 15, 54, 69, 96, 100],
    'two': [1, 1, 2, 2, 3, 3],
    'three': [6, 24, 15, 60, 45, 36]
})
df
```

```python
df['one'].unique() # returns unique values in column 'one'
```

```python
df['one'].nunique() # returns the number of unique values instead
```

```python
df['two'].value_counts() # returns the frequency of values
```

```python
df['one'].sum() # a basic function
```

```python
df['one'].apply(lambda x : x * 2) # applies a function
```

```python
# You can also apply a built-in function
```

```python
df.drop('two', axis = 1) # drops that column, 'inplace' is present
```

```python
df.columns # a list of column names
```

```python
df.index # a list of indexes
```

```python
df.sort_values('three') # sorts the column
# df.sort_values(by = 'three') same thing
```

```python
df.isnull() # returns boolean values indicating whether a value is null
```
