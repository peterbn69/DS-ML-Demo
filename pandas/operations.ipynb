```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    'one': [12, 15, 54, 69, 96, 100],
    'two': [1, 1, 2, 2, 3, 3],
    'three': [6, 24, 15, 60, 45, 36]
})
print(df)
```

```python
print(df['one'].unique()) # prints unique values in column 'one'
print(df['one'].nunique()) # prints the number of unique values instead
```

```python
print(df['two'].value_counts()) # prints the frequency of values
```

```python
print(df['one'].sum()) # a basic function

print(df['one'].apply(lambda x : x * 2)) # applies a function
# You can also apply a built-in function
```

```python
print(df.drop('two', axis = 1)) # drops that column, 'inplace' is present
```

```python
print(df.columns) # a list of column names
print(df.index) # a list of indexes
```

```python
print(df.sort_values('three')) # sorts the column
# df.sort_values(by = 'three') same thing
```

```python
print(df.isnull()) # returns boolean values indicating whether a value is null
```
