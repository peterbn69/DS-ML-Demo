# Pandas - Missing Data

```python
import numpy as np
import pandas as pd
```

```python
dic = {
    'A': [1, 2, np.nan],
    'B': [5, np.nan, np.nan],
    'C': [1, 2, 3]
}
```

```python
df = pd.DataFrame(dic)
print(df)
```

```python
print(df.dropna()) # drops ANY rows containing missing values, `inplace` is present
print()
print(df.dropna(axis = 1)) # drops ANY columns containing missing values
print()
print(df.dropna(thresh = 2)) # requires that rows with at least 2 non-NA values not be dropped
```

```python
print(df.fillna(value = 'TBD')) # replaces NA values with 'TBD', `inplace` is also present
```

```python
print(df['A'].fillna(value = df['A'].mean())) # replaces NA values in column A with the mean
```
