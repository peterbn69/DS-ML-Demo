# Pandas - Missing Data

```python
import numpy as np
import pandas as pd
```

```python
dic = {
    'A': [1, 2, np.nan],
    'B': [5, np.nan, np.nan],
    'C': [1, 2, 3]
}
```

```python
df = pd.DataFrame(dic)
df
```

```python
df.dropna() # drops ANY rows containing missing values, `inplace` is present
```

```python
df.dropna(axis = 1) # drops ANY columns containing missing values
```

```python
df.dropna(thresh = 2) # requires that rows with at least 2 non-NA values not be dropped
```

```python
df.fillna(value = 'TBD') # replaces NA values with 'TBD', `inplace` is also present
```

```python
df['A'].fillna(value = df['A'].mean()) # replaces NA values in column A with the mean
```
