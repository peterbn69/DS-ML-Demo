```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Company': ['Google', 'Google', 'Microsoft', 'Microsoft', 'Meta', 'Meta'],
    'Person': ['Sam', 'Charlie', 'Amy', 'Vanessa', 'Carl', 'Sarah'],
    'Sales': [200, 120, 340, 124, 243, 350]
}
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
print(df.groupby('Company')['Sales'].mean()) # groups by Company and returns the means of Sales
print()
print(df.groupby('Company')['Sales'].mean().loc['Microsoft']) # returns the mean Sales of Microsoft
```

```python
print(df.groupby('Company').count()) # now it works for the other columns
```

```python
# Other aggregate functions include max(), mean(), std(),...
```

```python
print(df.groupby('Company')['Sales'].describe())
print()
print(df.groupby('Company')['Sales'].describe().transpose())
```
