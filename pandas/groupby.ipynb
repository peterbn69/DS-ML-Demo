```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Company': ['Google', 'Google', 'Microsoft', 'Microsoft', 'Meta', 'Meta'],
    'Person': ['Sam', 'Charlie', 'Amy', 'Vanessa', 'Carl', 'Sarah'],
    'Sales': [200, 120, 340, 124, 243, 350]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
df.groupby('Company')['Sales'].mean() # groups by Company and returns the means of Sales
```

```python
df.groupby('Company')['Sales'].mean().loc['Microsoft'] # returns the mean Sales of Microsoft
```

```python
df.groupby('Company').count() # now it works for the other columns
```
