```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
    'first_name': ['Peter', 'Tony', 'Steve'],
    'last_name': ['Parker', 'Stark', 'Rogers'],
    'role': ['Spider-Man', 'Iron-Man', 'Captain America']
}, index = [0, 1, 2])
```

```python
df2 = pd.DataFrame({
    'first_name': ['Frank', 'Matthew'],
    'last_name': ['Castle', 'Murdock'],
    'role': ['The Punisher', 'Daredevil']
}, index = [3, 4])
```

```python
df3 = pd.DataFrame({
    'first_name': ['Bruce', 'Bruce'],
    'last_name': ['Banner', 'Wayne'],
    'role': ['The Hulk', 'Batman']
}, index = [5, 6])
```

```python
pd.concat([df1, df2, df3]) # concatenates data frames
```

```python
# You can concatenate along columns by specifying axis = 1
```

```python
left = pd.DataFrame({
    'first_name': ['Peter', 'Tony', 'Steve', 'Frank', 'Matthew', 'Bruce', 'Bruce'],
    'last_name': ['Parker', 'Stark', 'Rogers', 'Castle', 'Murdock', 'Banner', 'Wayne']
})
left.index.names = ['ID']
left
```

```python
right = pd.DataFrame({
    'role': ['Spider-Man', 'Iron-Man', 'Captain America', 'The Punisher', 'Daredevil', 'The Hulk', 'Batman']
})
right.index.names = ['ID']
right
```

```python
pd.merge(left, right, how = 'inner', on = 'ID') # merges two data frames on 'ID' (similar to SQL)
```

```python
# Joining is a convenient method for combining the columns of two data frames ON YOUR INDEX
```

```python
left = pd.DataFrame({
    'first_name': ['Peter', 'Tony', 'Steve', 'Frank', 'Matthew', 'Bruce', 'Bruce'],
    'last_name': ['Parker', 'Stark', 'Rogers', 'Castle', 'Murdock', 'Banner', 'Wayne']
})
right = pd.DataFrame({
    'role': ['Spider-Man', 'Iron-Man', 'Captain America', 'The Punisher', 'Daredevil', 'The Hulk', 'Batman']
})
left.join(right)
```
