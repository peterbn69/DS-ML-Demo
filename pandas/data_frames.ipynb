```python
from numpy.random import randn
np.random.seed(101)
```

```python
df = pd.DataFrame(randn(5, 4), ['A', 'B', 'C', 'D', 'E'], ['W', 'X', 'Y', 'Z'])
df
```

```python
df['W'] # df['W'] is a Series
```

```python
df[['W', 'Z']] # this is a DataFrame
```

```python
df['new'] = df['W'] + df['Y']
df['new'] # new column
```

```python
df.drop('new', axis = 1) # returns a column where 'new' is dropped, does not drop a column in place
```

```python
df.drop('new', axis = 1, inplace = True) # now it happens in place
df
```

```python
# SELECTING ROWS
```

```python
df.loc['A'] # select rows by labels
```

```python
df.iloc[0] # select rows by position
```

```python
df.loc['B', 'Y'] # row B col Y
```

```python
df.loc[['A', 'B'], ['W', 'Y']] # rows A B cols W Y
```

# Pandas - DataFrame Part 2

```python
df
```

```python
df > 0
```

```python
df[df > 0] # conditional selection
```

```python
df[df['W'] > 0] # returns only the filtered rows in col W
```

```python
df[df['W'] > 0]['X'] # only rows where W > 0 in column X
```

```python
df[df['W'] > 0][['X', 'Y']]
```

```python
df[(df['W'] > 0) & (df['Y'] > 1)] # multiple conditions, or will be |
```

```python
df.reset_index() # indexes reset to a column, inplace = True if changes happen in place
```

```python
new_idx = 'CA NY WY OR CO'.split() # neat trick
new_idx
```

```python
df['States'] = new_idx
df
```

```python
df.set_index('States') # indexes set to States, same for inplace = True
```

# Pandas - DataFrames Part 3

```python
# Index levels
outside = 'G1 G1 G1 G2 G2 G2'.split(' ')
inside = [1, 2, 3, 1, 2, 3]
hier_idx = list(zip(outside, inside))
hier_idx = pd.MultiIndex.from_tuples(hier_idx)
```

```python
hier_df = pd.DataFrame(randn(6, 2), hier_idx, ['A', 'B'])
hier_df
```

```python
hier_df.loc['G1']
```

```python
hier_df.loc['G1'].loc[1]
```

```python
hier_df.index.names = ['Groups', 'Numbers']
hier_df
```

```python
hier_df.loc['G2'].loc[1]['B']
```

```python
hier_df.xs(1, level = 'Numbers')
```
