# Pandas - DataFrames Part 1

```python
import numpy as np
import pandas as pd
```

```python
labels = np.array(['a', 'b', 'c'])
my_data = np.array([10, 20, 30])
dic = {
    'a': 10,
    'b': 20,
    'c': 30
}
```

```python
print(pd.Series(data = my_data)); print('\n')
print(pd.Series(data = my_data, index = labels)); print('\n')
print(pd.Series(dic))
```

```python
ser = pd.Series(dic)
print(ser)
```

```python
print(ser['a'])
ser['a'] = 50 # changeable
print(ser['a'])
```

```python
from numpy.random import randn
np.random.seed(101)
```

```python
df = pd.DataFrame(randn(5, 4), ['A', 'B', 'C', 'D', 'E'], ['W', 'X', 'Y', 'Z'])
print(df)
```

```python
print(df['W']) # df['W'] is a Series
print(df[['W', 'Z']]) # this is a DataFrame
```

```python
df['new'] = df['W'] + df['Y']
print(df['new']) # new column
print(df.drop('new', axis = 1)) # returns a column where 'new' is dropped, does not drop a column in place
```

```python
df.drop('new', axis = 1, inplace = True) # now it happens in place
print(df)
```

```python
# SELECTING ROWS
```

```python
print(df.loc['A']) # select rows by labels
print(df.iloc[0]) # select rows by position
```

```python
print(df.loc['B', 'Y']) # row B col Y
print(df.loc[['A', 'B'], ['W', 'Y']]) # rows A B cols W Y
```

# Pandas - DataFrame Part 2

```python
print(df)
print(df > 0)
```

```python
print(df[df > 0]) # conditional selection
print()
print(df[df['W'] > 0]) # prints only the filtered rows in col W
```

```python
print(df[df['W'] > 0]['X']) # only rows where W > 0 in column X
print(df[df['W'] > 0][['X', 'Y']])
```

```python
print(df[(df['W'] > 0) & (df['Y'] > 1)]) # multiple conditions, or will be |
```

```python
print(df.reset_index()) # indexes reset to a column, inplace = True if changes happen in place
```

```python
new_idx = 'CA NY WY OR CO'.split() # neat trick
print(new_idx)
```

```python
df['States'] = new_idx
print(df)
```

```python
print(df.set_index('States')) # indexes set to States, same for inplace = True
```

# Pandas - DataFrames Part 3

```python
# Index levels
outside = 'G1 G1 G1 G2 G2 G2'.split(' ')
inside = [1, 2, 3, 1, 2, 3]
hier_idx = list(zip(outside, inside))
hier_idx = pd.MultiIndex.from_tuples(hier_idx)
```

```python
hier_df = pd.DataFrame(randn(6, 2), hier_idx, ['A', 'B'])
print(hier_df)
```

```python
print(hier_df.loc['G1'])
print()
print(hier_df.loc['G1'].loc[1])
```

```python
hier_df.index.names = ['Groups', 'Numbers']
print(hier_df)
```

```python
print(hier_df.loc['G2'].loc[1]['B'])
```

```python
print(hier_df.xs(1, level = 'Numbers'))
```
