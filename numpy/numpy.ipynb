```python
# array indexing is similar to that of Python lists
```

```python
arr = np.arange(1, 11)
arr
```

```python
arr > 5 # array of boolean values
```

```python
bool_arr = arr > 5
arr[bool_arr] # conditional selection
```

```python
arr[arr > 5] # all in one
```

```python
# ---------- NUMPY OPERATIONS ----------
```

```python
arr = np.arange(0, 11)
# SCALAR OPERATIONS
print(arr + arr)
print(arr * arr)
print(arr - arr)
print(arr + 100)
```

```python
print(arr / arr) # nan for 0/0
print(1 / arr) # inf for 1/0
```

```python
print(np.sqrt(arr)) # square root
print(np.exp(arr)) # exponential
```
