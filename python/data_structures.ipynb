# Lists

```python
nums = [12, 15, 54, 69, 72, 84, 96] # = list((12, 15, 54, 69, 72, 84, 96))
```

```python
len(nums)
```

```python
54 in nums
```

```python
nums.insert(2, 24)
nums
```

```python
nums.append(100)
nums
```

```python
fruits = ["apple", "banana", "cherry"]
fruits.extend(["orange", "mango", "pineapple"]) # extend can be used with any iterable
fruits
```

```python
fruits.remove("mango") # a specific value
fruits
```

```python
fruits.pop(2) # an index, pop() will remove the last item
fruits
```

```python
nums.clear() # clear the list
nums
```

```python
for fruit in fruits:
    print(fruit, end=" ")
```

```python
[print(fruit) for fruit in fruits] # list comprehension
```

```python
" ".join(fruits) # join() method
```

```python
nums = [54, 12, 15, 96, 72, 84, 69]
nums.sort()
nums
```

```python
nums.sort(reverse = True)
nums
```

```python
nums.sort(key = lambda x : abs(x - 50)) # sort by distance from 50
nums
```

```python
nums.reverse() # reverse the list
nums
```

```python
nums = [12, 15, 54, 69, 72, 84, 96]
tmp = nums.copy() # or list(nums) or nums[:]
tmp
```

```python
nums += tmp # work just like the extend method
nums
```

```python
nums.count(12)
```

```python
nums.index(69)
```

# Tuples

```python
mytuple = ("apple", "banana", "cherry") # or = tuple(["apple", "banana", "cherry"])
mytuple
```

```python
len(mytuple)
```

```python
mytuple += ("orange", )
mytuple
```

```python
fruits = ("apple", "banana", "cherry")
(green, yellow, red) = fruits # unpacking

print(green)
print(yellow)
print(red)
```

```python
fruits = ("apple", "banana", "cherry", "strawberry", "raspberry")
(green, yellow, *red) = fruits

print(green)
print(yellow)
print(red)
```

```python
fruits = ("apple", "mango", "papaya", "pineapple", "cherry")
(green, *tropic, red) = fruits

print(green)
print(tropic)
print(red)
```

# Sets

```python
favs = {"banana", "pineapple", "cataloupe", "melon", "watermelon"} # or use the set() constructor
favs
```

```python
len(favs)
```

```python
favs.add("grapes") # add another element
favs
```

```python
favs.update({54, 69}) # add an iterable
favs
```

```python
favs.remove(54) # remove an element, but will raise an error if it does not exist
```

```python
favs.discard(69) # remove an element silently
```

```python
# pop() removes a random item, so it's safe not to use it
favs
```

```python
favs.clear()
favs # still present, use `del` to free memory
del favs
```

```python
even = {0, 2, 4, 6, 8}
odd = {1, 3, 5, 7, 9}
uni = even | odd # UNION, even.union(odd), can be used with iterables
uni
```

```python
empty = even & odd # INTERSECTION, even.intersection(odd), can be used with iterables
empty
```

```python
still_even = even - odd # DIFFERENCE, even.difference(odd), can be used with iterables
still_even
```

`symmetric_difference()` is equivalent to `^`

`intersection_update()` is equivalent to `&=`

`union_update()` is equivalent to `|=`

So on and so forth...

Other methods include:
- `copy()` returns a shallow copy
- `isdisjoint()`
- `issubset()` or `< | <=`
- `issuperset()` or `> | >=`

# Dictionaries

```python
myself = {
    "name": "Peter",
    "age": 19,
    "major": "Computer Science"
} # dict() constructor can also be used
myself
```

```python
myself["name"] # or myself.get("name")
```

```python
print(myself.items())
print(myself.keys())
print(myself.values())
```

```python
myself["date-of-birth"] = "January 5th, 2007" # add a new item, equivalent to myself.update({"date-of-birth": ...})
myself
```

```python
myself["something"] = "something"
myself.pop("something") # remove an item
myself
```

```python
myself["something"] = "something"
myself.popitem() # remove the last inserted item
myself
```

```python
myself.clear()
myself # still present
```

```python
del myself # now gone in memory
```

```python
myself = {
    "name": "Peter",
    "age": 19,
    "major": "Computer Science"
}

for key in myself: # myself.keys() by default
    print(key, end=" ")
print()

for value in myself.values():
    print(value, end=" ")
print()

for item in myself.items():
    print(item, end=" ")
```

```python
x = ('key1', 'key2', 'key3')
y = 0

thisdict = dict.fromkeys(x, y)

print(thisdict)
```

```python
myself = {
    "name": "Peter",
    "age": 19,
    "major": "Computer Science"
}
print(myself.setdefault("age", 18))
print(myself.setdefault("occupation", "student"))
# setdefault(key, def_val) returns the value of the key (def_val if not exist)
```
