# Iterators
An iterator is an object that contains a countable number of values.

An iterator consists of the methods `__iter()__` and `__next()__`.

```python
mytuple = ("apple", "banana", "cherry")
myit = iter(mytuple)

print(next(myit))
print(next(myit))
print(next(myit))
```

# Exceptions
`try`, `except`, `else`, `finally`

```python
a = int(input("a: "))
b = int(input("b: "))

try:
    print(b / a)
except ZeroDivisionError:
    print("Cannot divide by zero.")
else:
    print("No exception caught.")
finally:
    print("End of program.")
```

```python
try:
    raise Exception
except Exception:
    print("Exception has been raised and caught.")
```
