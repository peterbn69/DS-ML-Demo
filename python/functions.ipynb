# Functions

```python
def greet():
    print("Greetings!")
greet()
```

```python
def m_to_cm(n):
    return n * 100
m_to_cm(5)
```

```python
def do_nothing():
    pass # no statements, for future implementation
```

```python
def full_name(fname: str, lname: str = "Parker") -> str: # for readability, "Parker" is the default value
    return fname + " " + lname
full_name("Peter", "Parker")
```

```python
full_name("Peter")
```

```python
full_name(lname = "Parker", fname = "Peter")
```

```python
def greet(name, /): # positional arguments only
    print(f"Hello, {name}!")
greet("Peter")
greet("Bao")
```

```python
def greet(*, fname, lname): # keyword arguments only
    print(f"Hello, {fname} {lname}!")
greet(fname = "Peter", lname = "Parker")
```

```python
def give_candy(*kids): # *args
    print(f"I have given my candies to {", ".join(kids)}")
give_candy("Peter", "Frank", "Bruce")
```

```python
def introduce(**self): # **kwargs
    print(f"Hello, my name is {self["name"]} and I am {self["age"]} years old.")
introduce(name = "Peter", age = 19)
```

If you want to combine `*args` and `**kwarg`, the order must be regular parameter, `*args`, `**kwarg`.

```python
kids = ["Peter", "Frank", "Bruce"]
myself = {
    "name": "Peter",
    "age": 19
}
give_candy(*kids) # unpack an interable
introduce(**myself) # unpack a dictionary
```

# Lambdas

Syntax: `lambda arguments : expression`

```python
add_5 = lambda x : x + 5
add_5(5)
```

## Usage with Built-in Functions

```python
nums = [1, 2, 3, 4, 5]
doubled = list(map(lambda x : x * 2, nums))
doubled
```

```python
nums = [1, 2, 3, 4, 5, 6, 7, 8]
odd_nums = list(filter(lambda x: x % 2 != 0, nums))
odd_nums
```

# Generators

Generators are functions that can pause and resume their execution.

When a generator function is called, it returns a generator object, which is an iterator.

The code inside the function is not executed yet, it is only compiled. The function only executes when you iterate over the generator.

```python
def my_generator():
    yield 1
    yield 2
    yield 3

for value in my_generator():
    print(value, end=" ")
```

```python
def count_up_to(n):
    for i in range(n + 1):
        yield i

for i in count_up_to(5):
    print(i, end=" ")
```

Generators are memory-efficient because they generate values on-the-fly instead of storing everything in memory.

```python
def large_sequence(n):
    for i in range(n):
        yield i

# This doesn't create a million numbers in memory
gen = large_sequence(1000000)
print(next(gen))
print(next(gen))
print(next(gen))
```

```python
# List comprehension - creates a list
list_comp = [x * x for x in range(5)]
print(list_comp)

# Generator expression - creates a generator
gen_exp = (x * x for x in range(5))
print(gen_exp)
print(list(gen_exp))
```

```python
def hello():
    while True:
        name = yield # accept the sent value
        print(f"Hello {name}!")

gen = hello()
next(gen) # Prime the generator (MUST-DO)
gen.send("Peter")
gen.send("Bao")
gen.close() # Stop the generator
```
