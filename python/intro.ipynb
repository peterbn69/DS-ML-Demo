## Data Types
- Text Type: `str`
- Numeric Types: `int`, `float`, `complex`
- Sequence Types: `list`, `tuple`, `range`
- Mapping Type: `dict`
- Set Types: `set`, `frozenset`
- Boolean Type: `bool`
- Binary Types: `bytes`, `bytearray`, `memoryview`
- None Type: `None`

```python
x = 69 # int
y = 6.9 # float
z = 6 + 9j # complex

print(type(x))
print(type(y))
print(type(z))
```

```python
print(6 < 9)
print(6 == 9)
print(6 > 9)
```

Most values are evaluated to `True` if it has content. The following evaluates to `False`:
- `False`
- `""`
- `0`
- `(), [], {}`
- `None`

An object whose `__len__()` returns `0` is `False`

## Conditional Statements

```python
age = 19
if age <= 12:
    print("Child")
elif age <= 18:
    print("Teenager")
else:
    print("Adult")
```

```python
age = 20
print("Ineligible" if age < 18 else "Eligible")
```

Logical Operators: `and`, `or`, `not`

```python
if True:
    pass # do nothing
```

```python
day = 5
match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
    case _:
        print("Invalid")
```

```python
day = 5
match day:
    case 1 | 2 | 3 | 4 | 5:
        print("Weekday")
    case 6 | 7:
        print("Weekend")
    case _:
        print("Invalid")
```
