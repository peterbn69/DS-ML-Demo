```python
import seaborn as sns
%matplotlib inline
tips = sns.load_dataset("tips")
flights = sns.load_dataset("flights")
```

```python
tips
```

```python
tc = tips[["total_bill", "tip", "size"]].corr()
tc
```

```python
sns.heatmap(tc, annot=True)
```

```python
sns.heatmap(tc, annot=True, cmap="coolwarm")
```

```python
flights
```

```python
fp = flights.pivot_table(index="month", columns="year", values="passengers") # aggfunc for the aggregate function
fp
```

```python
sns.heatmap(fp, cmap="magma", linewidths=0.25, linecolor="violet")
```
