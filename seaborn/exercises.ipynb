```python
import seaborn as sns
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
sns.set_style("whitegrid")
titanic = sns.load_dataset("titanic")
titanic
```

```python
sns.jointplot(x="fare", y="age", data=titanic)
```

```python
sns.displot(titanic["fare"], color="red", bins=30)
```

```python
sns.boxplot(x="class", y="age", data=titanic, hue="class", legend=False, palette="rainbow")
```

```python
sns.swarmplot(x="class", y="age", data=titanic, hue="class", legend=False, palette="rainbow")
```

```python
sns.countplot(x="sex", data=titanic)
```

```python
sns.heatmap(titanic[["survived", "pclass", "age", "sibsp", "parch", "fare"]].corr(), cmap="coolwarm")
plt.title("titanic.corr()")
```

```python
g= sns.FacetGrid(data=titanic, col="sex")
g.map(plt.hist, "age")
```
