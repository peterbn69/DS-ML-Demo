```python
import seaborn as sns
import pandas as pd
```

```python
%matplotlib inline
# to see our visualizations in the notebook
```

```python
from pyodide.http import open_url

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
tips = pd.read_csv(open_url(url))
tips.head()
```

# Distribution Plots

```python
sns.displot(tips["total_bill"])
```

```python
sns.displot(tips["total_bill"], kde=True, bins=20)
```

```python
sns.jointplot(x="total_bill",y="tip",data=tips)
```

```python
sns.jointplot(x="total_bill",y="tip",data=tips, kind="hex")
```

```python
sns.jointplot(x="total_bill",y="tip",data=tips, kind="reg")
```

```python
sns.pairplot(tips, hue="sex") # also palette="coolwarm" for color
```

```python
sns.rugplot(tips["total_bill"])
```

```python
sns.kdeplot(tips["total_bill"])
```

# Categorical Plots

```python
import numpy as np
```

```python
sns.barplot(x="sex", y="total_bill", data=tips)
```

```python
sns.barplot(x="sex", y="total_bill", data=tips, estimator=np.std)
```

```python
sns.countplot(x="sex", data=tips)
```

```python
sns.boxplot(x="day", y="total_bill", data=tips)
```

```python
sns.boxplot(x="day", y="total_bill", data=tips, hue="smoker")
```

```python
sns.violinplot(x="day", y="total_bill", data=tips)
```

```python
sns.violinplot(x="day", y="total_bill", data=tips, hue="sex", split=True)
```

```python
sns.stripplot(x="day", y="total_bill", data=tips, hue="sex")
```

```python
sns.swarmplot(x="day", y="total_bill", data=tips)
```

```python
sns.violinplot(x="day", y="total_bill", data=tips)
sns.swarmplot(x="day", y="total_bill", data=tips, color="yellow")
```
