```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
%matplotlib inline
```

# Part 1

```python
x = np.linspace(0, 5, 11)
y = x**2
```

```python
x
```

```python
y
```

## 1. Functional

```python
plt.plot(x, y, color='red')
plt.xlabel('X Label')
plt.ylabel('Y Label')
plt.title('Title')
# plt.show() if not in Jupyter notebook
```

```python
plt.subplot(1, 2, 1) # nrows - ncols - current plot
plt.plot(x, y, color='green')
plt.subplot(1, 2, 2)
plt.plot(y, x, color='blue')
```

## 2. Object-Oriented

```python
fig = plt.figure() # creates a Figure object

axes = fig.add_axes([0.1, 0.1, 0.8, 0.8]) # left - bottom - width - height
axes.plot(x, y)
axes.set_xlabel('X Label')
axes.set_ylabel('Y Label')
axes.set_title('Set Title')
```

```python
fig = plt.figure()

axes1 = fig.add_axes([0.1, 0.1, 0.8, 0.8])
axes2 = fig.add_axes([0.2, 0.5, 0.4, 0.3])

axes1.plot(x, y)
axes1.set_title('BIGGER PLOT')

axes2.plot(y, x)
axes2.set_title('SMALLER PLOT')
```

# Part 2

```python
fig, axes = plt.subplots(nrows=1, ncols=2)

axes[0].plot(x, y)
axes[0].set_title('First Plot')

axes[1].plot(y, x)
axes[1].set_title('Second Plot')

plt.tight_layout() # deals with overlapping issues
```

`axes` is just a list of matplotlib axes, and it can be iterated through.

## Figure Size and DPI

```python
fig = plt.figure(figsize=(4, 2)) # height - width 

ax = fig.add_axes([0, 0, 1, 1])
ax.plot(x, y)
```

```python
fig,axes = plt.subplots(nrows=2, ncols=1, figsize=(8, 4))

axes[0].plot(x, y)

axes[1].plot(y, x)

plt.tight_layout()
```

```python
# fig.savefig('my_figure.png', dpi=200)
```

```python
fig = plt.figure()

ax = fig.add_axes([0, 0, 1, 1])
ax.plot(x, x**2, label='x^2')
ax.plot(x, x**3, label='x^3')

ax.legend(loc=0) # best location
# ax.legend(loc=(0.5, 0.5))
```

# Part 3

```python
fig = plt.figure()

ax = fig.add_axes([0, 0, 1, 1])

ax.plot(x, y, c='purple', lw=2, alpha=0.5, ls='--', marker='o', ms=5, mfc='blue') # or c = RGB Hex Code, alpha controls transparency
```

```python
fig = plt.figure()

ax = fig.add_axes([0, 0, 1, 1])

ax.plot(x, y, color='purple', lw=2, ls='--')
ax.set_xlim([0, 1])
ax.set_ylim([0, 2])
```

```python
from random import sample

fig, axes = plt.subplots(nrows=1, ncols=3, figsize=(8, 3))

x = np.arange(1, 11)
y = x**2
axes[0].scatter(x, y)

data = sample(range(1, 1000), 100)
axes[1].hist(data)

data = [np.random.normal(0, std, 100) for std in range(1, 4)]
axes[2].boxplot(data, vert=True, patch_artist=True)

fig.tight_layout()
```
